```python
import os
import io
import json
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import balanced_accuracy_score, confusion_matrix
from sklearn.preprocessing import StandardScaler

warnings.filterwarnings("ignore")
pd.set_option("display.max_columns", 30)
np.random.seed(42)

OUT = Path("project02_real_outputs")
OUT.mkdir(exist_ok=True)
DATA = Path("cryptic_data")
DATA.mkdir(exist_ok=True)
```

```python
import urllib.request

ZENODO_RECORD = "15679886"
BASE = f"https://zenodo.org/records/{ZENODO_RECORD}/files"

FILES = {"GENOMES.parquet":          "GENOMES.parquet", "UKMYC_PHENOTYPES.parquet": "UKMYC_PHENOTYPES.parquet",
    "ENA_LOOKUP.parquet":       "ENA_LOOKUP.parquet", "SITES.csv.gz":             "SITES.csv.gz",
    "DRUG_CODES.csv.gz":        "DRUG_CODES.csv.gz",}

def download_if_needed(name, local_name):
    local_path = DATA / local_name
    if local_path.exists():
        print(f"✓ already present: {local_path}")
        return local_path
    url = f"{BASE}/{name}?download=1"
    print(f"Downloading {name} …")
    try:
        urllib.request.urlretrieve(url, local_path)
        print(f"  → saved {local_path} ({local_path.stat().st_size/1e6:.1f} MB)")
        return local_path
    except Exception as e:
        print(f"  ✗ failed: {e}")
        return None

paths = {}
for remote, local in FILES.items():
    paths[local] = download_if_needed(remote, local)

print("\nDownload step finished.")
```

```python
genomes = pd.read_parquet(paths["GENOMES.parquet"])
phenos  = pd.read_parquet(paths["UKMYC_PHENOTYPES.parquet"])

print("GENOMES", genomes.shape, "index:", genomes.index.names)
print("PHENOTYPES", phenos.shape, "index:", phenos.index.names)

phenos = phenos.reset_index()
genomes = genomes.reset_index()

print("\nUnique drugs:", sorted(phenos["DRUG"].unique()))
print("BINARY_PHENOTYPE counts:\n", phenos["BINARY_PHENOTYPE"].value_counts(dropna=False))
print("\nWGS_PREDICTION_STRING examples:")
print(genomes["WGS_PREDICTION_STRING"].dropna().head(8).tolist())
```

```python
print("Records before join")
print(f"  GENOMES: {len(genomes):,}")
print(f"  UKMYC_PHENOTYPES: {len(phenos):,}")

df = phenos.merge(genomes, on="UNIQUEID", how="inner", suffixes=("_pheno", "_geno"))

n_before = len(phenos)
n_after  = len(df)
lost = n_before - n_after
print(f"\nAfter inner join on UNIQUEID")
print(f"  Retained: {n_after:,}")
print(f"  Lost: {lost:,}  ({100*lost/n_before:.1f}%)")

df = df[df["DRUG"].isin(["INH", "RIF"])].copy()
df = df[df["BINARY_PHENOTYPE"].isin(["S", "R"])].copy()
df["resistant"] = (df["BINARY_PHENOTYPE"] == "R").astype(int)

print(f"\nFinal INH+RIF labelled set: {len(df):,} pairs")
print(df.groupby("DRUG")["resistant"].agg(["count", "mean"]).round(3))

df.to_csv(OUT / "cryptic_inh_rif_labelled.csv", index=False)
```

```python
known_markers = pd.DataFrame({
    "gene":  ["katG", "rpoB", "inhA", "pncA", "embB", "gyrA", "rpsL", "rrs"],
    "drug":  ["INH",  "RIF",  "INH",  "PZA",  "EMB",  "FQ",   "STM",  "AMI"],
    "start": [2153889, 759807, 1674202, 2288681, 4246514, 7302, 781561, 1471846],
    "classic": ["S315T", "RRDR (S450L etc.)", "promoter", "LoF", "M306V/I",
                "A90V/D94G", "K43R/K88R", "A1401G"]
})
print(known_markers.to_string(index=False))
known_markers.to_csv(OUT / "h37rv_resistance_loci.csv", index=False)
```

```python
def parse_prediction_string(s):
    """Return list of 0/1/None for each character (R=1, S=0, U/other=None)."""
    if not isinstance(s, str) or len(s) < 5:
        return []
    s = s.replace(" ", "")
    out = []
    for ch in s:
        if ch == "R":
            out.append(1)
        elif ch == "S":
            out.append(0)
        else:
            out.append(None)
    return out

parsed = df["WGS_PREDICTION_STRING"].apply(parse_prediction_string)
max_len = parsed.str.len().max()
print(f"Prediction string length (max): {max_len}")

geno_feats = pd.DataFrame(parsed.tolist(), index=df.index, columns=[f"geno_pos_{i}" for i in range(max_len)])

for c in geno_feats.columns:
    geno_feats[c + "_missing"] = geno_feats[c].isna().astype(int)
    geno_feats[c] = geno_feats[c].fillna(0).astype(int)

print("Genomic feature matrix:", geno_feats.shape)
print(geno_feats.head(3))

meta_subset = df[["PHENOTYPE_QUALITY", "PLATEDESIGN"]].astype(str).fillna("UNKNOWN")
meta_feats = pd.get_dummies(meta_subset, drop_first=True)
meta_feats["is_RIF"] = (df["DRUG"] == "RIF").astype(int)

X = pd.concat([geno_feats, meta_feats], axis=1)
y = df["resistant"].values
meta = df[["UNIQUEID", "DRUG", "BINARY_PHENOTYPE", "resistant", "WGS_PREDICTION_STRING"]].reset_index(drop=True)
X = X.reset_index(drop=True)

print("Final X shape:", X.shape)
print("Resistance prevalence:", y.mean().round(3))
```

```python
from sklearn.metrics import balanced_accuracy_score

def best_position_for_drug(drug):
    mask = meta["DRUG"] == drug
    best_pos, best_bal = None, -1
    for i in range(max_len):
        col = f"geno_pos_{i}"
        if col not in X.columns:
            continue
        pred = X.loc[mask, col].values
        valid = X.loc[mask, col + "_missing"].values == 0
        if valid.sum() < 100:
            continue
        bal = balanced_accuracy_score(y[mask][valid], pred[valid])
        if bal > best_bal:
            best_bal = bal
            best_pos = i
    return best_pos, best_bal

print("Finding best genomic position for each drug (catalogue baseline)...")
pos_inh, bal_inh = best_position_for_drug("INH")
pos_rif, bal_rif = best_position_for_drug("RIF")
print(f"INH best position: {pos_inh}  (balanced acc ≈ {bal_inh*100}%)")
print(f"RIF best position: {pos_rif}  (balanced acc ≈ {bal_rif*100}%)")

def catalogue_pred(row, pos_inh, pos_rif):
    s = str(row["WGS_PREDICTION_STRING"]).replace(" ", "")
    if row["DRUG"] == "INH" and pos_inh is not None and len(s) > pos_inh:
        return 1 if s[pos_inh] == "R" else 0
    if row["DRUG"] == "RIF" and pos_rif is not None and len(s) > pos_rif:
        return 1 if s[pos_rif] == "R" else 0
    return 0

meta = meta.copy()
meta["baseline_pred"] = meta.apply(lambda r: catalogue_pred(r, pos_inh, pos_rif), axis=1)
meta["baseline_correct"] = meta["baseline_pred"] == meta["resistant"]

print("\nCatalogue baseline (real genomic calls)")
for drug in ["INH", "RIF"]:
    sub = meta[meta["DRUG"] == drug]
    print(f"  {drug}: accuracy {sub['baseline_correct'].mean()*100}%  (n={len(sub)})")
```

```python
results = {}

for drug in ["INH", "RIF"]:
    mask = (meta["DRUG"] == drug).values
    Xd = X[mask].fillna(0)
    yd = y[mask]

    print(f"\n{drug} (n={len(yd)}, resistance={yd.mean():.3f})")

    clf = RandomForestClassifier( n_estimators=200, max_depth=12, class_weight={0: 1.0, 1: 5.0}, random_state=42, n_jobs=-1,)

    cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
    y_pred = cross_val_predict(clf, Xd, yd, cv=cv)

    bal = balanced_accuracy_score(yd, y_pred)
    cm  = confusion_matrix(yd, y_pred, labels=[0, 1])
    tn, fp, fn, tp = cm.ravel()
    vme = fn / (fn + tp) if (fn + tp) else 0.0
    me  = fp / (tn + fp) if (tn + fp) else 0.0

    print(f"Balanced accuracy : {bal*100}%")
    print(f"Very Major Error  : {vme*100}%   (Resistant -> Susceptible)")
    print(f"Major Error       : {me*100}%   (Susceptible -> Resistant)")
    print("Confusion matrix (rows=true, cols=pred):\n", cm)

    clf.fit(Xd, yd)
    results[drug] = { "model": clf, "balanced_accuracy": bal, "vme": vme, "me": me, "cm": cm, "feature_names": list(Xd.columns),}

print("\nML vs real catalogue baseline")
for drug in ["INH", "RIF"]:
    base_acc = meta.loc[meta.DRUG == drug, "baseline_correct"].mean()
    print(f"{drug}:  ML bal.acc {results[drug]['balanced_accuracy']*100:.1f}%   "
          f"Catalogue {base_acc*100:.1f}%")
```
